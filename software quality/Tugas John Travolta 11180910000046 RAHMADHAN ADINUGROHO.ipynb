```python
jam = int(input('Masukan Jam Kerja : '))
print ('Jam Kerja ' + str(jam))
```

```python
if 0 <= jam <= 40:
    gaji = jam * 15000
    print ('Gaji adalah ' + str(gaji))
else:
    lebih = (jam - 40)
    gaji1 = lebih * 15000 * 1.5
    gaji2 = 40 * 15000
    gaji = gaji1 + gaji2
    print ('Gaji adalah ', int(gaji))
```

```python
pengeluaran = int(input('Masukan Jumlah Pengeluaran : '))
print ('Jumlah pengeluaran adalah ', int(pengeluaran))
```

```python
if pengeluaran == gaji:
    print('Tidak Bisa Menabung')
elif pengeluaran > gaji:
    print('Cari Pemasukan Lagi')
else:
    nabung = gaji - pengeluaran
    print('Anda dapat Menabung Sejumlah ', int(nabung))
```
